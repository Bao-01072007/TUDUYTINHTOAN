### W7A1
Viết hàm tìm kiếm nhị phân trên một danh sách đã được sắp xếp tăng dần. Trả về chỉ số của phần tử hoặc -1 nếu không tìm thấy. Mẫu hàm: defbinary_search(arr, left, right, target):


```python
def defbinary_search(arr, target):
    if not arr: # Nếu danh sách trống
        return None
    left = 0
    right = len(arr)-1
    while left <= right:  # Do đây là dãy tăng
        mid = (left + right)//2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target: # Do dãy tăng dần nên nếu giá trị ở giữa mà nhỏ hơn giá trị cần tìm thì chỉ cần tìm nửa bên phải của dãy thôi
            left = mid + 1
        else:
            right = mid - 1
    return -1
arr = list(map(int, input().split()))
target = int(input())
print(defbinary_search(arr, target))


            
```

```python
def binary_search(arr, left, right, target):
    if left > right:
        return -1
    mid = (left + right)//2
    if arr[mid] == target:
        return mid
    elif arr[mid] > target:
        return binary_search(arr, left, mid-1, target)
    else:
        return binary_search(arr, mid+1, right, target)
arr = [1, 3, 5, 7, 9, 11]
result = binary_search(arr, 0, len(arr) - 1, 7)
print(result)
```

### W7A2
Cho danh sách arr và một số nguyên x, đếm xem x xuất hiện bao nhiêu lần trong danh sách. def count_occurrences(arr, x) -> int:



```python
def count_occurrences(arr, x):
    dem = 0
    for i in arr:
        if i not in arr:
            return 0
        else:
            if i == x:
                dem += 1
    return dem

arr = list(map(int, input().split()))
x = int(input())
print(count_occurrences(arr, x))

```

### W7A3
Viết hàm thực hiện thuật toán bubble sort, Selection Sort để sắp xếp danh sách theo thứ tự tăng dần. Sắp xếp nổi bọt (Bubble Sort);


```python
lst = list(map(int, input().split()))
lst.sort()
print(lst)
```

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_index = i # Coi vị trí hiện tại là nhỏ nhất
        for j in range(i+1, n): # Tìm các vị trí nhỏ hơn
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr
arr = list(map(int, input().split()))
print(selection_sort(arr))
```

### W7A4
Viết hàm tìm phần tử có số lần xuất hiện nhiều nhất trong danh sách. Nếu có nhiều phần tử cùng số lần, trả về phần tử xuất hiện trước.


```python
def xuat_hien_nhieu_nhat(lst):
    # Đếm số lần xuất hiện bằng dict
    dem = {}
    for i in lst:
        if i in dem:
            dem[i] += 1  # Số lần xuất hiện của i tăng thêm 1
        else:
            dem[i] = 1
    xuat_hien_max = 0
    for i in dem:
        if dem[i] > xuat_hien_max:
            xuat_hien_max = dem[i]
    for i in lst:
        if dem[i] == xuat_hien_max:
            return i
lst = list(map(int, input().split()))
print(xuat_hien_nhieu_nhat(lst))
```

### W7A16
Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.



```python
def check_mang_dep(arr):
    ds_dem = {}
    for i in arr:
        if i in ds_dem:
            ds_dem[i] += 1
        else:
            ds_dem[i] = 1
    # kiểm tra xem có phải bảng đẹp hay không
    bang_dep = True
    for x, dem in ds_dem.items():
        if x != dem:
            bang_dep = False
            print("Không là bảng đẹp")
            break
    if bang_dep:
        return "Là bảng đẹp"
    # Tính số lượng cần xóa
    y = 0
    for x, dem in ds_dem.items(): # Duyệt từng phần tử trong dict phải dùng .items
        if x == 0: # Không thể xuất hiện 0 lần khi số 0 đã xuất hiện
            y += dem
        elif dem >= x:
            y += dem - x
        else:
            y += dem
    return y
arr = list(map(int, input().split()))
print(check_mang_dep(arr))



```

### W7A17
Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore.
Nếu không có phần tử nào thỏa mãn thì không in gì


```python
def tim_so(arr):
    n =len(arr)
    so_can_tim = None  # Ban đầu là chưa có gì
    dem = 0
    # Sẽ coi phần tử x là số cần tìm, nếu tiếp theo gặp x tiếp thì sẽ tăng thêm 1, không thì sẽ trừ đi 1
    # Số tồn tại cuối cùng và thỏa mãn lớn hơn n/2 lần thì sẽ thỏa mãn
    for x in arr:
        if dem == 0:
            so_can_tim = x
            dem = 1
        elif x == so_can_tim:
            dem += 1
        else:
            dem -= 1
    # Kiểm tra xem số tồn tại cuối cùng đó có thỏa mãn n/2 không
    dem_thuc_te = 0
    for x in arr:
        if x == so_can_tim:
            dem_thuc_te += 1
    if dem_thuc_te > n/2:
        return so_can_tim
    return None
arr = list(map(int, input().split()))
print(tim_so(arr))
```

### W7A19
Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không
dùng bất cứ thuật toán sắp xếp nào


```python
def merge_2_mang(arr1, arr2):
    # Cho 2 mảng cùng về dãy tăng
    if len(arr1) > 1 and arr1[0] > arr1[-1]:
        arr1 = arr1[::-1]
    if len(arr2) > 1 and arr2[0] > arr2[-1]:
        arr2 = arr2[::-1]
    s = []
    i = 0 # Cho con trỏ cho arr1
    j = 0 # Cho con trỏ cho arr2
    while i < len(arr1) and j < len(arr2): # Một trong 2 mảng mà chạy hết các phần tử rồi thì dừng vòng lặp
        if arr1[i] < arr2[j]: # So sánh 2 thằng đầu tiên của mảng, ai bé hơn cho vào s
           s.append(arr1[i])
           i += 1
        else:
            s.append(arr2[j])
            j += 1
    # Nếu phần tử của mảng mà vẫn còn dư thì thêm vào cuối s
    # Dùng extend để thêm từng cái vào phía sau
    # Dùng append để thêm nguyên khối phía sau
    s.extend(arr1[i:])
    s.extend(arr2[j:])
    return s
arr1 = list(map(int, input().split()))
arr2 = list(map(int, input().split()))
print(merge_2_mang(arr1, arr2))
     
```

### W7A25
Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số



```python
s = input()
k = int(input())
n = len(s)
# Số chữ số cần xóa
can_xoa = n-k
stack = []  # Tạo 1 ngăn xếp rỗng
for i in s:
    # Đáp ứng 3 điều kiện:
    # 1. Ngăn xếp không rỗng
    # 2. Chữ số hiện tại lớn hơn chữ số cuối cùng trong stack
    # 3. Vẫn còn lượt để xóa bỏ chữ số (can_xoa >0)
    while stack and stack[-1] < i and can_xoa > 0:
        stack.pop() # Xóa phần tử cuối cùng của stack
        can_xoa -= 1 # Sau khi xóa thì sẽ giảm dần can_xoa
    stack.append(i)
# Sau khi duyệt hết 1 lượt mà vẫn còn lượt xóa
# Ta sẽ cắt phần phía cuối
print("".join(stack[:k]))
```
